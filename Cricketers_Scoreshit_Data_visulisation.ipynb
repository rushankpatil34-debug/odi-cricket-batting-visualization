```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

```

```python
df= pd.read_csv("odb.csv")
df.head()
```

```python
top10 = df.sort_values(by="Runs", ascending= False).head(10)
sns.barplot(data = top10,x= "Runs",y = "Player")
plt.xticks(rotation=90)
plt.title("Top 10 Players Highest Runs in ODI Cricket")
plt.grid()
plt.show()

topave10 = df.sort_values(by="Ave", ascending= False).head(10)
sns.barplot(data = topave10,x= "Player",y = "Ave")
plt.xticks(rotation=90)
plt.ylabel("Average")
plt.grid()
plt.title("Top 10 Players Highest Average in ODI Cricket")
plt.show()

top5 = df.sort_values(by="HS", ascending= False).head(10)
sns.boxplot(data = top5,x= "Player",y = "HS")
plt.xticks(rotation=90)
plt.title("Top 5 Players Highest score in ODI Cricket")
plt.grid()
plt.show()

# top12 = df.sort_values(by="SR", ascending= False)
sns.histplot(data = df,x= "SR", kde = True)
plt.xticks(rotation=90)
plt.title("Strikre Rate Distribution")
plt.grid()
plt.show()

sns.regplot(data = df,x= "Ave",y = "Runs", scatter_kws={"s": 60}, line_kws={"linewidth": 2})
plt.xticks(rotation=90)
plt.ylabel("Total Run")
plt.xlabel("Average")
plt.grid()
plt.title("Relationship between batting average and runs")
plt.show()

sns.jointplot(data= df,x="Runs",y="SR",kind= "scatter")
plt.show()

cols= ["Runs","Ave","SR","100","50"]
sns.pairplot(df[cols],kind="reg")
plt.show()

df["Century_Group"] = pd.cut(df["100"],bins=[-1,10,30,60], labels=["Low","Medium","High"])
sns.scatterplot(data= df, x= "Ave", y="Runs", hue="Century_Group",size="SR")
plt.title("Run Vs average by Century Group")
plt.grid()
plt.show()

sns.barplot(data= df, x= "6s", y="4s")
plt.show()


Top10= df.nlargest(10, "Runs")
plot_data= Top10[["Player","4s","6s"]].melt(id_vars="Player",var_name="Boundary",value_name="Count")
sns.barplot(data= plot_data, x="Player",y= "Count",hue= "Boundary")
plt.title("Top 10 Players: 4s vs 6s")
plt.tight_layout()
plt.xticks(rotation=90)
plt.gca().invert_yaxis()
plt.show()
```
